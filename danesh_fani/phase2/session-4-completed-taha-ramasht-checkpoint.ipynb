<div class="homework-fa" dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:1.9;font-family:Vazirmatn,'Noto Sans Arabic',Tahoma,sans-serif;font-size:16px"><p dir="rtl" style="direction:rtl;text-align:right">حدس: ۳ دما بالاتر از ۳۰ هستند.</p></div>

```python

temperatures = [24, 31, 29, 35, 22, 33, 30, 26]
print("Number of readings:", len(temperatures))
```

```python

print("Minimum temperature:", min(temperatures))
print("Maximum temperature:", max(temperatures))
mean_temp = sum(temperatures) / len(temperatures)
print("Average temperature: {:.2f}".format(mean_temp))

above_30 = 0
for temp in temperatures:
    if temp > 30:
        above_30 += 1
print("Temperatures above 30:", above_30)

above_mean = 0
for temp in temperatures:
    if temp > mean_temp:
        above_mean += 1
print("Temperatures above average:", above_mean)
```

<div class="homework-fa" dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:1.9;font-family:Vazirmatn,'Noto Sans Arabic',Tahoma,sans-serif;font-size:16px"><p dir="rtl" style="direction:rtl;text-align:right">حدس: ۲ نفر بالاترین نمره را گرفته‌اند.</p></div>

```python

scores_by_name = {
    "Ali": 14,
    "Sara": 18,
    "Reza": 9,
    "Mina": 18,
    "Nima": 11,
    "Parisa": 16,
}
```

```python

scores = list(scores_by_name.values())
average_score = sum(scores) / len(scores)
print("Class average:", round(average_score, 2))

above_average_names = []
for name, score in scores_by_name.items():
    if score > average_score:
        above_average_names.append(name)
print("Above average:", above_average_names)

top_score = max(scores)
top_names = []
for name, score in scores_by_name.items():
    if score == top_score:
        top_names.append(name)
print("Top score:", top_score)
print("Students with top score:", top_names)

passed_count = 0
for score in scores:
    if score >= 10:
        passed_count += 1
print("Scores greater than or equal to 10:", passed_count)
```

<div class="homework-fa" dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:1.9;font-family:Vazirmatn,'Noto Sans Arabic',Tahoma,sans-serif;font-size:16px"><h2 dir="rtl" style="direction:rtl;text-align:right">۳. داده‌های ناقص</h2><p dir="rtl" style="direction:rtl;text-align:right">دما برای بعضی دستگاه‌ها ثبت نشده است.</p><ol dir="rtl" style="padding-right:1.5em"><li>نام دستگاه‌های بدون دما را پیدا کنید. از <code dir="ltr" style="direction:ltr;unicode-bidi:isolate">get(&quot;temp&quot;)</code> استفاده کنید.</li><li>دماهای موجود را در یک لیست تازه بریزید؛ داده‌های اصلی را تغییر ندهید.</li><li>میانگین دماهای موجود را حساب کنید.</li><li>نام و تعداد دستگاه‌هایی را چاپ کنید که دمایشان بیشتر از ۴۰ است.</li></ol></div>

<div class="homework-fa" dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:1.9;font-family:Vazirmatn,'Noto Sans Arabic',Tahoma,sans-serif;font-size:16px"><p dir="rtl" style="direction:rtl;text-align:right">حدس: دمای ۴ دستگاه قابل‌استفاده است.</p></div>

```python

devices = [
    {"name": "A", "temp": 35, "voltage": 220},
    {"name": "B", "temp": 42, "voltage": 218},
    {"name": "C", "voltage": 224},
    {"name": "D", "temp": None, "voltage": 219},
    {"name": "E", "temp": 48, "voltage": 221},
    {"name": "F", "temp": 39, "voltage": 222},
]
```

```python

missing_devices = []
available_temps = []

for device in devices:
    if device.get("temp") is None:
        missing_devices.append(device["name"])
    else:
        available_temps.append(device["temp"])

print("Devices with missing temperatures:", missing_devices)
print("Available temperatures:", available_temps)

if len(available_temps) > 0:
    average_temp = sum(available_temps) / len(available_temps)
    print("Average available temperature:", round(average_temp, 2))
else:
    print("There is no temperature data.")

hot_names = []
for device in devices:
    if device.get("temp") is not None and device["temp"] > 40:
        hot_names.append(device["name"])

print("Devices above 40:", hot_names)
print("Count above 40:", len(hot_names))
```

<div class="homework-fa" dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:1.9;font-family:Vazirmatn,'Noto Sans Arabic',Tahoma,sans-serif;font-size:16px"><p dir="rtl" style="direction:rtl;text-align:right">تعداد دستگاه‌ها ۶ تاست، اما فقط ۴ دما ثبت شده است. برای دستگاه C مقدار temp وجود ندارد و برای دستگاه D مقدار آن None است؛ پس این دو دستگاه دمای قابل استفاده ندارند.</p></div>

<div class="homework-fa" dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:1.9;font-family:Vazirmatn,'Noto Sans Arabic',Tahoma,sans-serif;font-size:16px"><h2 dir="rtl" style="direction:rtl;text-align:right">۴. دو تابع کاربردی</h2><ol dir="rtl" style="padding-right:1.5em"><li>تابع <code dir="ltr" style="direction:ltr;unicode-bidi:isolate">safe_average(numbers)</code> را بنویسید: میانگین لیست را برگرداند و برای لیست خالی <code dir="ltr" style="direction:ltr;unicode-bidi:isolate">None</code> بدهد.</li><li>تابع <code dir="ltr" style="direction:ltr;unicode-bidi:isolate">count_above(numbers, threshold)</code> تعداد مقادیر بزرگ‌تر از آستانه را برگرداند.</li><li>هر دو تابع را با داده‌های زیر امتحان کنید. میانگین را با <code dir="ltr" style="direction:ltr;unicode-bidi:isolate">statistics.mean</code> هم مقایسه کنید.</li><li>فرق <code dir="ltr" style="direction:ltr;unicode-bidi:isolate">return</code> و <code dir="ltr" style="direction:ltr;unicode-bidi:isolate">print()</code> چیست؟</li></ol></div>

<div class="homework-fa" dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:1.9;font-family:Vazirmatn,'Noto Sans Arabic',Tahoma,sans-serif;font-size:16px"><p dir="rtl" style="direction:rtl;text-align:right">حدس: چون لیست خالی است، تابع باید None برگرداند.</p></div>

```python

def safe_average(numbers):
    if len(numbers) == 0:
        return None
    return sum(numbers) / len(numbers)


def count_above(numbers, threshold):
    count = 0
    for number in numbers:
        if number > threshold:
            count += 1
    return count


test_values = [20, 30, 31, 35]
empty_values = []

print("Average:", safe_average(test_values))
print("Average of empty list:", safe_average(empty_values))
print("Count above 30:", count_above(test_values, 30))

import statistics
print("Average with statistics.mean:", statistics.mean(test_values))
```

```python

students = [
    {"name": "Sara", "hours": 8, "attendance": 95, "previous_grade": 18, "needs_support": False},
    {"name": "Ali", "hours": 2, "attendance": 60, "previous_grade": 11, "needs_support": True},
    {"name": "Mina", "hours": 6, "attendance": 85, "previous_grade": 16, "needs_support": False},
    {"name": "Reza", "hours": 1, "attendance": 55, "previous_grade": 9, "needs_support": True},
    {"name": "Nima", "hours": 4, "attendance": 80, "previous_grade": 14, "needs_support": True},
]
```

```python

X = []
y = []

for student in students:
    row = [student["hours"], student["attendance"], student["previous_grade"]]
    X.append(row)
    y.append(student["needs_support"])

print("Length of X:", len(X))
print("Length of y:", len(y))
print("Third row of X:", X[2])
print("Third value of y:", y[2])
```

```python

inspection_records = [
    {"name": "A", "temp": 35, "voltage": 220, "inspection_required": False},
    {"name": "B", "temp": 42, "voltage": 218, "inspection_required": True},
    {"name": "C", "temp": 39, "voltage": 224, "inspection_required": True},
    {"name": "D", "temp": 48, "voltage": 219, "inspection_required": False},
    {"name": "E", "temp": None, "voltage": 220, "inspection_required": True},
    {"name": "F", "temp": 31, "voltage": 221, "inspection_required": False},
    {"name": "G", "temp": 41, "voltage": 217, "inspection_required": True},
    {"name": "H", "temp": 37, "voltage": 222, "inspection_required": False},
    {"name": "I", "voltage": 219, "inspection_required": False},
    {"name": "J", "temp": 44, "voltage": 220, "inspection_required": True},
]
```

```python

missing_names = []
valid_records = []

for record in inspection_records:
    if record.get("temp") is None:
        missing_names.append(record["name"])
    else:
        valid_records.append(record)

print("Devices with missing temperatures:", missing_names)
print("Total devices:", len(inspection_records))
print("Valid devices:", len(valid_records))


valid_temps = []
for record in valid_records:
    valid_temps.append(record["temp"])

if len(valid_temps) > 0:
    print("Minimum temperature:", min(valid_temps))
    print("Maximum temperature:", max(valid_temps))
    print("Average temperature:", round(sum(valid_temps) / len(valid_temps), 2))
else:
    print("There is no valid temperature data.")


def rule_check(temp):
    return temp > 40

agreement_count = 0
different_names = []

for record in valid_records:
    rule_result = rule_check(record["temp"])
    human_result = record["inspection_required"]
    print(record["name"], record["temp"], rule_result, human_result)

   
    if rule_result == human_result:
        agreement_count += 1
    else:
        different_names.append(record["name"])

agreement_percent = agreement_count / len(valid_records) * 100
print("Agreements:", agreement_count)
print("Different cases:", different_names)
print("Agreement percentage:", round(agreement_percent, 2), "%")


X = []
y = []

for record in valid_records:
    X.append([record["temp"], record["voltage"]])
    y.append(record["inspection_required"])

print("X:", X)
print("y:", y)
print("Same number of rows:", len(X) == len(y))
```
